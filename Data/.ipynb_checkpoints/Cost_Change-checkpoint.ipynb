# Partial Dependency Plot as Gentrification Measure

The premiums that homeowners pay to buy into different neighborhoods change over time for a variety of reasons. For investors looking to enter a market, there are several reasons to be interested in this. For consumers looking to buy a home for themselves this measure can indicate which neighborhoods are up-and-coming. For contractors this can guide what type of house they decide to build; an area undergoing gentrification will have more demand for higher-end homes in the future. 

In this notebook we consider a third application of this information. We consider the case of two hypothetical investors, one who bought a house in 2023 in Licoln Park (the neighborhood with the highest premium) vs an investor who considered the change in the partial dependence plot from 2021-2022 to 2023 and bought a house in North Center. These neighboorhoods are very close to each other, even sharing a boarder.

```python
import numpy as np
import pandas as pd
import xgboost as xgb
import json
```

```python
data = pd.read_parquet(r"../Data/sale_and_crime_data/chicago_sales_with_crime.parquet")
```

```python
features_to_exclude = ["pin","is_multisale",'flood_fema_sfha', "sale_filter_same_sale_within_365", "sale_filter_less_than_10k", "single_v_multi_family", "zip_code",
                      "longitude", "latitude", "sale_date_parsed", "class", 'centroid_x_crs_3435', 'centroid_y_crs_3435', 'pin10', 'nearest_metra_route_dist_ft', 'nearest_new_construction_pin10',
                       "school_elementary_district_name",'nearest_golf_course_dist_ft', "school_secondary_district_name", "township_code",  "neighborhood_code",'sale_filter_deed_type', 'nearest_cta_route_dist_ft',
                                 'nearest_vacant_land_pin10','airport_noise_dnl','attic_finish','census_acs5_tract_geoid', 'nearest_neighbor_1_pin10','nearest_neighbor_2_pin10','nearest_neighbor_3_pin10','sale_date_parsed','sale_date', 'sale_year']
data = data.drop(columns = features_to_exclude)
```

```python
cat_cols = ["type_of_residence", "construction_quality", "garage_size",
            "basement_type", "ext_wall_material", "repair_condition", "basement_finish",
            "central_heating", "central_air", "community_area"]

for col in cat_cols:
    data[col] = data[col].astype("category")
```

```python
def XGBModel(df, hyperparams):
    y=df["sale_price"]
    features = list(df.columns)
    features.remove("sale_price")
    X=df[features]
    
    model = xgb.XGBRegressor(
        random_state=42, 
        enable_categorical=True, 
        tree_method="hist",
        **hyperparams
    )
    model.fit(
        X,y,
        verbose=False
    )
    return model
```

Now that we have the basic code we will need to train our models, we carve out from the data the housing prices from 2021 to 2023.

```python
data_first = data[data["year"]<2023]
```

```python
data_second = data[data["year"]==2023]
```

```python
with open(r"..\notebooks\hyperparameters.json", "r") as f:
    hyperparams = json.load(f)
```

```python
model_pre_2023 = XGBModel(data_first, hyperparams)
model_2023 = XGBModel(data_second, hyperparams)
```

```python
def categorical_pdp(model, X, feature, categories=None):
    """
    Compute partial dependence (marginal expected value) for a categorical feature.
    
    model: fitted XGBRegressor (enable_categorical=True)
    X: DataFrame used for prediction (same columns/dtypes as training)
    feature: name of the categorical column
    categories: optional list/order of categories to evaluate; defaults to all observed categories
    """
    if categories is None:
        categories = X[feature].cat.categories if hasattr(X[feature], "cat") else X[feature].unique()

    results = []
    X_temp = X.copy()

    for cat in categories:
        X_temp[feature] = cat
        # preserve categorical dtype/categories so xgboost doesn't complain
        X_temp[feature] = X_temp[feature].astype(X[feature].dtype)
        preds = model.predict(X_temp)
        results.append({"category": cat, "mean_prediction": preds.mean()})

    return pd.DataFrame(results).sort_values("mean_prediction", ascending=False)
```

```python
X1=data_first.drop(columns="sale_price")
X2=data_second.drop(columns="sale_price")
```

```python
Change_in_pdp = categorical_pdp(model_2023, X2, "community_area")
Original = categorical_pdp(model_pre_2023, X1, "community_area")
Change_in_pdp["mean_prediction"] = Change_in_pdp["mean_prediction"]- Original["mean_prediction"]
```

```python
Change_in_pdp = Change_in_pdp.sort_values("mean_prediction", ascending = False)
```

```python
print(Change_in_pdp.to_string(index=False))
```

From the above we see that North Center (community area 5) is one of the highest increases in housing premiums. We will assume that the two investors by average houses in their neighborhoods in 2024.

```python
Licoln_buy_price = data[(data["year"]==2024) & (data["community_area"]==7)]["sale_price"].mean()
print(Licoln_buy_price)
```

```python
NC_buy_price = data[(data["year"]==2024) & (data["community_area"]==5)]["sale_price"].mean()
print(NC_buy_price)
```

We will suppose that the two investors also sell their property at the average value of the neighborhoods in 2025.

```python
Licoln_profit = data[(data["year"]==2025) & (data["community_area"]==7)]["sale_price"].mean() - Licoln_buy_price
NC_profit = data[(data["year"]==2025) & (data["community_area"]==5)]["sale_price"].mean() - NC_buy_price

print(f"The profit of our first investor who bought in Licoln Park is ${Licoln_profit:.2f}")
print('-'*100)
print(f"The profit of our second investor who bought in North Center is ${NC_profit:.2f}")
```
